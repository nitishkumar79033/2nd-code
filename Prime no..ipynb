```python
#prime no.
num = int(input("Enter a number: "))
if num <= 1:
    print("not a prime no.")
else :
    for i in range(2,num):
        if num % i == 0:
            print("not a prime no.")
            break
    else:
        print("prime no.")
```

```python
#Even or Odd number 
num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```

```python
#Palindrome number
num = int(input("Enter a number: "))
temp = num
rev = 0
while num > 0:
    dig = num % 10
    rev = rev * 10 + dig
    num = num // 10
if temp == rev:
    print("Palindrome number")
else:
    print("Not a palindrome number")
```
