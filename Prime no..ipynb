```python
#prime no.
num = int(input("Enter a number: "))
if num <= 1:
    print("not a prime no.")
else :
    for i in range(2,num):
        if num % i == 0:
            print("not a prime no.")
            break
    else:
        print("prime no.")
```

```python
#Even or Odd number 
num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```
